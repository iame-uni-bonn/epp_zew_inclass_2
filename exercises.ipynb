````python
def newtons_method(func, derivative, x_start):
    x = x_start
    for _ in range(10):
        x = x - 2 * func(x) / derivative(x)
    return x


def func(x):
    return x**2


def derivative(x):
    return 2 * x


newtons_method(func, derivative, x_start=1)
````

### Question 21

For the remainder of the course, it is important that you know how to execute Python 
code in several formats, using the correct environment. 

Please execute the following code in as many formats as you can, using the correct pixi
environment:

- a notebook in vscode
- a .py file in vscode
- a .py file using the terminal.

Notes:
- You need to create a new Python file for this. Please commit this file.
- Both the  notebook and the .py file must live in the same directory as
  `pyproject.toml`; for the last option, you need to run the script from that same
  directory.
- The reason is that the code will check whether the Python interpreter is the one from
  the local pixi environment and raise an error otherwise.

````python
import sys
from pathlib import Path


def check_python_interpreter_is_local_pixi_default():
    """Check that the local pixi environment with name 'default' is active."""
    expected = Path(".pixi/envs/default/bin/python").resolve()
    interpreter_path = Path(sys.executable).resolve()

    if expected != interpreter_path:
        print(
            "You ran the code using the wrong Python interpreter. You used:"
            f"\n\n{interpreter_path}\n\n"
            "Please run the code using the local pixi environment at:"
            f"\n\n{expected}\n\n"
        )
    else:
        print(
            "You have successfully run this code in the local pixi environment.\n\n"
            "Great job!"
        )


check_python_interpreter_is_local_pixi_default()
````
